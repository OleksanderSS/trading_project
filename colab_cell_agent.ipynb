```python
!pip install -U huggingface_hub transformers accelerate gradio

from huggingface_hub import login
from transformers import AutoModelForCausalLM, AutoTokenizer, pipeline
import gradio as gr

HF_TOKEN = "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"
login(token=HF_TOKEN)

# CodeLlama-Instruct (без use_auth_token)
codellama_tokenizer = AutoTokenizer.from_pretrained("codellama/CodeLlama-7b-Instruct-hf", token=HF_TOKEN)
codellama_model = AutoModelForCausalLM.from_pretrained("codellama/CodeLlama-7b-Instruct-hf", token=HF_TOKEN)
codellama_pipe = pipeline("text-generation", model=codellama_model, tokenizer=codellama_tokenizer)

# Mistral-Instruct
mistral_tokenizer = AutoTokenizer.from_pretrained("mistralai/Mistral-7B-Instruct-v0.2", token=HF_TOKEN)
mistral_model = AutoModelForCausalLM.from_pretrained("mistralai/Mistral-7B-Instruct-v0.2", token=HF_TOKEN)
mistral_pipe = pipeline("text-generation", model=mistral_model, tokenizer=mistral_tokenizer)

def audit_code(code):
    prompt = f"Audit this Python code. Find bugs, weak points, and suggest improvements:\n{code}"
    return codellama_pipe(prompt, max_length=512)[0]["generated_text"]

def explain_architecture(code):
    prompt = f"Explain the architecture of this code. Suggest better logic and reasoning:\n{code}"
    return mistral_pipe(prompt, max_length=512)[0]["generated_text"]

iface = gr.Interface(
    fn=[audit_code, explain_architecture],
    inputs="text",
    outputs=["text","text"],
    title="Code Auditor + Architecture Assistant",
    description="CodeLlama для аудиту, Mistral для пояснень"
)

iface.launch(share=True)
```
